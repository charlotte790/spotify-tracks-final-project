## Weiye Part 9: Random Forest Baseline

This section reproduces the same modeling setup used in the basic-model notebook: the selected Spotify audio features, track_genre as the target, a GroupShuffleSplit by track_id to avoid leakage, and StandardScaler for consistency.

```python
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    f1_score,
    precision_score,
    recall_score,
)
from sklearn.model_selection import GroupShuffleSplit, RandomizedSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

RANDOM_STATE = 42

selected_features = [
    "danceability",
    "energy",
    "key",
    "loudness",
    "mode",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo",
    "time_signature",
]


def prepare_modeling_data():
    df = pd.read_csv(DATA_PATH)

    unnamed_columns = [col for col in df.columns if str(col).startswith("Unnamed")]
    if unnamed_columns:
        df = df.drop(columns=unnamed_columns)

    df_clean = df.drop_duplicates().copy()
    df_clean = df_clean.dropna(subset=["track_id", "track_genre"] + selected_features)

    X = df_clean[selected_features].copy()
    y = df_clean["track_genre"].copy()
    groups = df_clean["track_id"].copy()

    group_split = GroupShuffleSplit(
        n_splits=1,
        test_size=0.2,
        random_state=RANDOM_STATE,
    )
    train_index, test_index = next(group_split.split(X, y, groups=groups))

    X_train = X.iloc[train_index].copy()
    X_test = X.iloc[test_index].copy()
    y_train = y.iloc[train_index].copy()
    y_test = y.iloc[test_index].copy()
    groups_train = groups.iloc[train_index].copy()
    groups_test = groups.iloc[test_index].copy()

    scaler = StandardScaler()
    X_train_scaled = pd.DataFrame(
        scaler.fit_transform(X_train),
        columns=selected_features,
        index=X_train.index,
    )
    X_test_scaled = pd.DataFrame(
        scaler.transform(X_test),
        columns=selected_features,
        index=X_test.index,
    )

    overlap = set(groups_train).intersection(set(groups_test))

    print("Rows after cleaning:", df_clean.shape[0])
    print("Training rows:", X_train_scaled.shape[0])
    print("Test rows:", X_test_scaled.shape[0])
    print("Number of genre classes:", y.nunique())
    print("Overlapping track_id count:", len(overlap))

    return X_train_scaled, X_test_scaled, y_train, y_test


X_train_scaled, X_test_scaled, y_train, y_test = prepare_modeling_data()


advanced_results = []


def evaluate_predictions(model_name, y_true, y_pred):
    result = {
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision_weighted": precision_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0,
        ),
        "Recall_weighted": recall_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0,
        ),
        "F1_weighted": f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0,
        ),
        "F1_macro": f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0,
        ),
    }
    advanced_results.append(result)

    print("\n" + "=" * 80)
    print(f"Classification Report: {model_name}")
    print("=" * 80)
    print(classification_report(y_true, y_pred, zero_division=0))

    return result
```

```python
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

random_forest.fit(X_train_scaled, y_train)
random_forest_pred = random_forest.predict(X_test_scaled)

rf_result = evaluate_predictions(
    "Random Forest",
    y_test,
    random_forest_pred,
)

pd.DataFrame([rf_result])
```

## Weiye Part 11: Neural Network Baseline

This section trains a simple neural-network baseline on the same scaled features. It uses TensorFlow/Keras when available; if TensorFlow is not installed locally, it falls back to scikit-learn's MLPClassifier so the notebook can still run.

```python
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)
n_classes = len(label_encoder.classes_)

try:
    import tensorflow as tf
    from tensorflow.keras.callbacks import EarlyStopping
    from tensorflow.keras.layers import Dense, Dropout
    from tensorflow.keras.models import Sequential

    tf.random.set_seed(RANDOM_STATE)

    neural_network = Sequential(
        [
            Dense(128, activation="relu", input_shape=(X_train_scaled.shape[1],)),
            Dropout(0.3),
            Dense(64, activation="relu"),
            Dropout(0.3),
            Dense(32, activation="relu"),
            Dense(n_classes, activation="softmax"),
        ]
    )

    neural_network.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )

    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True,
    )

    history = neural_network.fit(
        X_train_scaled,
        y_train_encoded,
        validation_split=0.2,
        epochs=50,
        batch_size=256,
        callbacks=[early_stopping],
        verbose=1,
    )

    nn_probabilities = neural_network.predict(X_test_scaled)
    nn_pred_encoded = np.argmax(nn_probabilities, axis=1)
    neural_network_method = "TensorFlow/Keras"

except ImportError:
    from sklearn.neural_network import MLPClassifier

    print("TensorFlow is not installed. Using sklearn MLPClassifier fallback.")

    neural_network = MLPClassifier(
        hidden_layer_sizes=(128, 64, 32),
        activation="relu",
        solver="adam",
        alpha=0.0001,
        batch_size=256,
        learning_rate="adaptive",
        max_iter=80,
        early_stopping=True,
        validation_fraction=0.2,
        n_iter_no_change=5,
        random_state=RANDOM_STATE,
        verbose=True,
    )

    neural_network.fit(X_train_scaled, y_train_encoded)
    nn_pred_encoded = neural_network.predict(X_test_scaled)
    neural_network_method = "sklearn MLPClassifier"

nn_pred = label_encoder.inverse_transform(nn_pred_encoded)

nn_result = evaluate_predictions(
    "Neural Network",
    y_test,
    nn_pred,
)

print("Neural network implementation used:", neural_network_method)
pd.DataFrame([nn_result])
```

## Weiye Part 12: Random Forest Hyperparameter Tuning

This section tunes only the Random Forest model using RandomizedSearchCV, which is more practical than a full grid search for this larger multi-class dataset.

```python
 rf_param_distributions = {
    "n_estimators": [100, 200, 300],
    "max_depth": [10, 20, 30, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
}

rf_for_tuning = RandomForestClassifier(
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

rf_random_search = RandomizedSearchCV(
    estimator=rf_for_tuning,
    param_distributions=rf_param_distributions,
    n_iter=10,
    cv=3,
    scoring="f1_macro",
    random_state=RANDOM_STATE,
    n_jobs=-1,
    verbose=2,
)

rf_random_search.fit(X_train_scaled, y_train)

print("Best Random Forest parameters:")
print(rf_random_search.best_params_)
print("Best cross-validation F1_macro:", rf_random_search.best_score_)

tuned_random_forest = rf_random_search.best_estimator_
tuned_rf_pred = tuned_random_forest.predict(X_test_scaled)

tuned_rf_result = evaluate_predictions(
    "Tuned Random Forest",
    y_test,
    tuned_rf_pred,
)

pd.DataFrame([tuned_rf_result])
```

## Top-3 Accuracy Evaluation

Top-3 Accuracy checks whether the correct genre is among the model's three most confident predictions. This is useful for the 114-class Spotify genre task because exact top-1 prediction is very strict.

```python
from pathlib import Path
import numpy as np
import pandas as pd

def calculate_top3_accuracy(y_true, probabilities, class_labels):
    y_true_array = np.asarray(y_true)
    probabilities = np.asarray(probabilities)
    class_labels = np.asarray(class_labels)

    top3_indices = np.argsort(probabilities, axis=1)[:, -3:]
    top3_labels = class_labels[top3_indices]

    return np.mean([true_label in predicted_labels for true_label, predicted_labels in zip(y_true_array, top3_labels)])


top3_results = []

# Random Forest
rf_for_top3 = globals().get("random_forest", globals().get("rf_model", None))

if rf_for_top3 is not None:
    rf_probabilities = rf_for_top3.predict_proba(X_test_scaled)
    rf_top3_accuracy = calculate_top3_accuracy(
        y_test,
        rf_probabilities,
        rf_for_top3.classes_
    )

    top3_results.append({
        "Model": "Random Forest",
        "Top3_Accuracy": rf_top3_accuracy
    })

    print(f"Random Forest Top-3 Accuracy: {rf_top3_accuracy:.4f}")
else:
    print("Random Forest model not found.")


# Neural Network
nn_for_top3 = globals().get("neural_network", globals().get("nn_model", None))

if nn_for_top3 is not None and "label_encoder" in globals():
    if "nn_probabilities" in globals():
        nn_probabilities_for_top3 = nn_probabilities
    elif hasattr(nn_for_top3, "predict_proba"):
        nn_probabilities_for_top3 = nn_for_top3.predict_proba(X_test_scaled)
    else:
        nn_probabilities_for_top3 = nn_for_top3.predict(X_test_scaled)

    nn_top3_accuracy = calculate_top3_accuracy(
        y_test,
        nn_probabilities_for_top3,
        label_encoder.classes_
    )

    top3_results.append({
        "Model": "Neural Network",
        "Top3_Accuracy": nn_top3_accuracy
    })

    print(f"Neural Network Top-3 Accuracy: {nn_top3_accuracy:.4f}")
else:
    print("Neural Network model or label_encoder not found.")


# Tuned Random Forest
tuned_rf_for_top3 = globals().get("tuned_random_forest", globals().get("best_rf", None))

if tuned_rf_for_top3 is not None:
    tuned_rf_probabilities = tuned_rf_for_top3.predict_proba(X_test_scaled)
    tuned_rf_top3_accuracy = calculate_top3_accuracy(
        y_test,
        tuned_rf_probabilities,
        tuned_rf_for_top3.classes_
    )

    top3_results.append({
        "Model": "Tuned Random Forest",
        "Top3_Accuracy": tuned_rf_top3_accuracy
    })

    print(f"Tuned Random Forest Top-3 Accuracy: {tuned_rf_top3_accuracy:.4f}")
else:
    print("Tuned Random Forest model not found.")


top3_accuracy_df = pd.DataFrame(top3_results)

RESULTS_DIR = Path(RESULTS_DIR)
RESULTS_DIR.mkdir(exist_ok=True)

top3_accuracy_path = RESULTS_DIR / "weiye_top3_accuracy_results.csv"
top3_accuracy_df.to_csv(top3_accuracy_path, index=False)

print("\nTop-3 Accuracy Results:")
display(top3_accuracy_df)

print("Saved Top-3 Accuracy results to:", top3_accuracy_path)
```
